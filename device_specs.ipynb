```python
import pandas as pd
import numpy as np

file_path = "ibm_fez_calibrations_2026-04-23T06_43_44Z.csv"
df = pd.read_csv(file_path)

# Extract 1D metrics
t1 = df['T1 (us)'].dropna()
t2 = df['T2 (us)'].dropna()
ro_error = df['Readout assignment error'].dropna()
sq_length = df['Single-qubit gate length (ns)'].dropna()
sq_error = df['√x (sx) error'].dropna()

# Extract 2Q metrics (parsing strings)
tq_lengths = []
for val in df['Gate length (ns)'].dropna():
    if isinstance(val, str) and val.strip():
        parts = val.split(';')
        for p in parts:
            if ':' in p:
                try:
                    tq_lengths.append(float(p.split(':')[1]))
                except:
                    pass

tq_errors = []
for val in df['CZ error'].dropna():
    if isinstance(val, str) and val.strip():
        parts = val.split(';')
        for p in parts:
            if ':' in p:
                try:
                    tq_errors.append(float(p.split(':')[1]))
                except:
                    pass

# Calculate stats
metrics = [
    ("T1 Time (us)", t1),
    ("T2 Time (us)", t2),
    ("Single-Qubit Gate Time (ns)", sq_length),
    ("Two-Qubit Gate Time (ns)", tq_lengths),
    ("Readout Error", ro_error),
    ("Single-Qubit Gate Error", sq_error),
    ("Two-Qubit Gate Error", tq_errors)
]

for name, data in metrics:
    if len(data) > 0:
        d = np.array(data)
        print(f"{name}|{np.min(d)}|{np.median(d)}|{np.mean(d)}|{np.max(d)}")
    else:
        print(f"{name}|No data")
```
